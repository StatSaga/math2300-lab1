SymPy does exact arithmetic, so answers appear as fractions instead of decimals.

```python
A_demo = np.array([[1, 2], [3, 4]])
b_demo = np.array([[5], [6]])

aug = sp.Matrix(np.hstack([A_demo, b_demo]))   # augmented matrix [A | b]
print("[A | b] =")
print(aug)

R, pivots = aug.rref()                          # reduced row echelon form
print("rref of [A | b] =")
print(R)
```

## Part 3 – Finding the Inverse of a Matrix

To find $C^{-1}$ by hand, we row reduce the augmented matrix $[C\,|\,I]$. When the left side becomes $I$, the right side is $C^{-1}$. Here is the $4\times 4$ matrix we will use:

$$C=\begin{pmatrix}1&1&1&-1\\-1&-1&0&1\\2&2&1&-1\\1&3&0&0\end{pmatrix}$$

### Python syntax
Two new tools: slicing out the right half of a matrix, and asking whether two matrices agree.

```python
C = np.array([[1, 1, 1, -1],
              [-1, -1, 0, 1],
              [2, 2, 1, -1],
              [1, 3, 0, 0]])

# A small example with a 2x2 matrix (not C)
D = np.array([[2, 1], [1, 1]])
aug_D = sp.Matrix(np.hstack([D, np.eye(2, dtype=int)]))   # [D | I]
R_D, _ = aug_D.rref()
print("rref of [D | I] =")
print(R_D)

Dinv_rref = R_D[:, 2:]             # [rows, columns]: all rows, columns 2 and onward (the right half of a 2x4)
print("Right half (this is the inverse of D) =")
print(Dinv_rref)
```

```python
Dinv_np = np.linalg.inv(D)         # NumPy's inverse (decimals)
print("np.linalg.inv(D) =")
print(Dinv_np)

# Are two matrices the same? Compare as decimals, allowing for tiny rounding error:
print("Same?", np.allclose(np.array(Dinv_rref, dtype=float), Dinv_np))
```

## Part 4 – When the Computer's Inverse Is Wrong

So far the computer's answers matched ours. Now we test a famous family of matrices, the **Hilbert matrices**, whose $(i,j)$ entry is

$$H_{ij}=\frac{1}{i+j-1}.$$

For example,
$$H_3=\begin{pmatrix}1&\tfrac12&\tfrac13\\ \tfrac12&\tfrac13&\tfrac14\\ \tfrac13&\tfrac14&\tfrac15\end{pmatrix}.$$

Every Hilbert matrix is invertible, and its inverse has integer entries. Run the next cell to define a function that builds $H_n$ in ordinary decimal (floating point) arithmetic.

```python
def hilbert(n):
    return np.array([[1/(i + j - 1) for j in range(1, n + 1)] for i in range(1, n + 1)])

def hilbert_exact(n):
    return sp.Matrix(n, n, lambda i, j: sp.Rational(1, i + j + 1))

np.set_printoptions(precision=4, suppress=False, linewidth=120)
hilbert(3)
```

### 4a. Does $H\,H^{-1}$ equal $I$?
The cell below computes `H @ np.linalg.inv(H)` and measures how far it is from $I$ (the largest entry of $HH^{-1}-I$, in absolute value). Run it, then **change `n`** to 4, 8, 10, 12, and 14.

```python
n = 4
H = hilbert(n)
Hinv = np.linalg.inv(H)
P = H @ Hinv

print("H @ inv(H) =")
print(np.round(P, 3))
print("\nLargest error from I:", np.abs(P - np.eye(n)).max())
```

### 4b. Table of errors

```python
print(" n     largest error in H*inv(H) - I")
for n in [3, 4, 6, 8, 10, 12, 14]:
    H = hilbert(n)
    err = np.abs(H @ np.linalg.inv(H) - np.eye(n)).max()
    print(f"{n:2d}     {err:.2e}")
```

### 4c. The consequence for solving systems
Choose the answer first: let $x=(1,1,\dots,1)$ and set $b=Hx$. Then we **know** the exact solution. See how well $x=H^{-1}b$ recovers it.

```python
print(" n     largest error in x = inv(H) b")
for n in [3, 6, 8, 10, 12, 14]:
    H = hilbert(n)
    x_true = np.ones(n)
    b = H @ x_true
    x_calc = np.linalg.inv(H) @ b
    print(f"{n:2d}     {np.abs(x_calc - x_true).max():.2e}")
```

### 4d. The exact answer
SymPy uses exact fractions rather than decimals. The function `hilbert_exact(n)` builds $H_n$ with exact fractions. Here is the pattern for $n=3$:

```python
H3 = hilbert_exact(3)
H3inv = H3.inv()
print("H3 * inv(H3) =")
print(H3 * H3inv)
print("Equals I?", H3 * H3inv == sp.eye(3))
```

**Exact arithmetic: always right, but with a cost**

SymPy stores every number as an exact fraction of whole numbers, such as $\tfrac{1}{13}$. Nothing is ever rounded, so every step of the row reduction is exactly correct. That is why $HH^{-1}=I$ comes out `True` for $n=12$ and $n=14$.

Decimal arithmetic can't store a number like $\tfrac13$ exactly. For matrices like $H$, those tiny rounding errors get magnified until the answer is wrong.

**The cost:** exact arithmetic never gives a wrong answer, but it gets slower and needs more memory as $n$ grows, because the numbers inside the computation keep getting longer. The largest entry of $H^{-1}$ has 19 digits at $n=14$ and 303 digits at $n=200$. Sooner or later a computer runs out of time or memory. Exact arithmetic also needs exact inputs, so it can't handle measured data or numbers like $\sqrt2$.

**In short:** decimal arithmetic is fast but can be wrong. Exact arithmetic is correct but slow, and it eventually becomes impractical too.
